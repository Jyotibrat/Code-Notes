```python
import pandas as pd # Importing the pandas library
import numpy as np # Importing the numpy library
```

```python
# This is the sample data of Stations through which the trains passes and what is the total time taken to complete the destination.

railway = {
    'Station': ["Guwahati - Jorhat", "Dibrugarh - Guwahati", "Mariani - New Delhi"],
    'Train': ["Jana-Shatabdi", "Nagaland Express", "Rajdhani"],
    'Time': [8, 12, 36]
}
```

```python
# Creating a Dataframe of the Data. A Dataframe is a collection of rows and columns in which we keep the data.

df = pd.DataFrame(railway)
```

```python
df
```

```python
#Storing the Dataframe in a .csv file.

df.to_csv('railway.csv')
```

```python
# read the .csv file and assigning it to a variable such that using that variable we can perform operations.

rail = pd.read_csv('railway.csv')
```

```python
# This will return the data in the 'Time' column. 

rail['Time']
```

```python
# This will return the data of the default index of the .csv file. 

rail['Unnamed: 0']
```

```python
# This will return the data at the index 1 of the column 'Station'.

rail['Station'][1]
```

```python
# This will change the data at the index 0 of the column 'Time'. The change is only on the outside like temporary and not saved in the .csv file. 

rail['Time'][0] = 7
```

```python
# Printing the content of the .csv file after changing the data.

rail
```

```python
# After changing the data, saving it in the .csv file. Now, the changed data will be saved in the .csv file.

rail.to_csv('railway.csv')
```

```python
# This will add a indexing layer to the pre existing indexing layer consisting of numbers.

rail.index = ['one', 'two', 'three']
```

```python
# Changing the data of the pre-indexing layer.

rail.index = ['ek', 'dui', 'tin']
```

```python
# Printing the data after altering the data.

rail
```
