```python
# Imported pandas and numpy library as pd and np respectively.

import pandas as pd
import numpy as np
```

```python
# Made a Dataframe of random numbers between 0 and 1 of 10 rows and 10 columns.

dataframe = pd.DataFrame(np.random.rand(10, 10))
```

```python
# Printing the top five rows along with all the respective columns.

dataframe.head()
```

```python
# Using this way can change all the values of the column of index '0' to None but, it is not preferred in normally because when we work on real datasets then sometimes the code becomes to long to detect the variable we are working on is a copy or making a view so while writting code with functions and working with dataset and we need to manipulate the dataset we must use .loc function or .iloc function.

dataframe[0] = None
dataframe
```

```python
# This function checks which values in the dataset is with null data type which is a boolean data type.

dataframe.isnull()
```

```python
# We are manipulating the dataset using the .loc function.

dataframe.loc[:,[0]] = 56
dataframe.isnull()
```

```python
# Creating a Dataframe consisting of 'names', 'toy' and 'born'.

df = pd.DataFrame({"name": ['Alfred', 'batman', 'Alfred'],
                        "toy": [np.nan, 'Batmobile', 'Bullwhip'],
                        "born": [pd.NaT, pd.Timestamp("1940-04-25"),
                        pd.NaT]})
df
```

```python
# This function removes any row which consist of the word "na" in any case (upper case or lower case).

df.dropna()
```

```python
# This argument explicitly removes the rows which consists of the word "na".

df.dropna(axis="rows")
```

```python
# Printing the dataframe.

df
```

```python
# This argument explicitly removes the columns which consists of the word "na".

df.dropna(axis="columns")
```

```python
# This removes only those rows which has all the elements consisting of the word "na" in a way or the other.

df.dropna(how="all")
```

```python
# This argument any rows whose any element consists of the word "na" in a way or the other. But, in this case we choose to remove the columns explicitly by writing the argument 'axis=1', normally by default it removes the rows and not columns. 

df.dropna(how="any", axis=1)
```

```python
df.drop_duplicates(subset='name')
```

```python
df.drop_duplicates(subset='name', keep='last')
```
