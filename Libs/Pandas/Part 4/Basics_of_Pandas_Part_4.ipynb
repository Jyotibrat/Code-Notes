```python
# Imported pandas and numpy library as pd and np respectively.

import pandas as pd
import numpy as np
```

```python
# Creating a Dataframe of random numbers with 500 rows and 10 columns.

dataframe_1 = pd.DataFrame(np. random.rand(500, 10))
```

```python
# Printing the Dataframe

dataframe_1
```

```python
# Creating a Dataframe of random numbers with 500 rows and 10 columns.

dataframe_2 = pd.DataFrame(np.random.random((500, 10)))
```

```python
# Printing the Dataframe

dataframe_2
```

```python
# Sorting the Dataframe on the basis of index. Here axis means columns and if ascending is True it means that indexing will be like 0 to ... and if false then vice versa. If axis is 1 then we sort the columns and if not then we do not sort and sorting is done using the 'ascending'.

dataframe_1.sort_index(axis=0, ascending=False)
```

```python
# Explanation above☝️

dataframe_1.sort_index(axis=1, ascending=False)
```

```python
# It is not copying. It is called as making a view of the Dataset. It is like making a pointer of the original dataset. The dataset is in the original memory location and it makes a pointer of it like we change the view then the original will also change.

dataframe = dataframe_2
```

```python
# It prints the series of the Dataframe which is the [0] index series for this case.

dataframe[0]
```

```python
# Changing the data at the index[0][0] to "Hello World".
dataframe[0][0] = "Hello World"
```

```python
# Printing the Dataframe.

dataframe
```

```python
# This means a copy and not a pointer like the data is stored in 2 different locations one is the copy and another one is original.

dataframe_copy = dataframe_2.copy()
```

```python
# we have changed the data at the index [0][0] of the copy of the dataframe and not the original dataframe.

dataframe_copy[0][0] = 12345
```

```python
# Printing the copied dataframe.

dataframe_copy
```

```python
# Printing the dataframe and we can notice that the data at the index [0][0] has not changed that means the data has only been changed of the copy dataframe.

dataframe_2
```

```python
# Changing the data at the index [0][0] using the loc function.

dataframe_2.loc[0, 0] = 6789
```

```python
# Printing the Dataframe.

dataframe_2
```

```python
# This changes the default index of the columns which are numbers to Alphabets.

dataframe_2.columns = list('ABCDEFGHIJ')
```

```python
# This will print the top four rows and along with their respective columns.

dataframe_2.head(4)
```

```python
# Changing the data at the index [0][0] using the loc function.

dataframe_2.loc[0, 0] = 1357
```

```python
# Printing the Dataframe.

dataframe_2
```

```python
# Changing the data at the index [0][0] using the loc function.

dataframe_2.loc[0, 'A'] = 2468
```

```python
# Printing the Dataframe.

dataframe_2
```

```python
# This will remove the '0' column but it will be temporarily that is there will be no effect in the memory location.

dataframe_2.drop(0, axis=1)
```

```python
# Printing the Dataframe.

dataframe_2
```

```python
# This will remove the '0' column but it will be permanent that is there will be full effect in the memory location.

dataframe_2 = dataframe_2.drop(0, axis=1)
```

```python
# This will remove the '(0, 'A')' column but it will be permanent that is there will be full effect in the memory location. If we need to do the same in the rows section then we need select '0' in the axis section.

dataframe_2 = dataframe_2.drop((0, 'A'), axis=1)
```

```python
# Printing the Dataframe.

dataframe_2
```
