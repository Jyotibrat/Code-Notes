```python
# Imported pandas and numpy library as pd and np respectively.

import pandas as pd
import numpy as np
```

```python
# Made a series of random 50 numbers using the pandas library.

series = pd.Series(np.random.rand(50))
```

```python
# Printing the series

series
```

```python
# This shows the datatype like it is a series or a dataframe.

type(series)
```

```python
# Making a Dataframe of random numbers of 500 rows and 10 columns using the pandas Library and also allocating the index of the Dataframe using the .arange() function which is similar to the range() function.

dataframe = pd.DataFrame(np.random.rand(500, 10), index = np.arange(1, 501))
```

```python
# Printing the Dataframe

dataframe
```
