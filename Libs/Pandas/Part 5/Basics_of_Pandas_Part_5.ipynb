```python
# Imported pandas and numpy library as pd and np respectively.

import pandas as pd
import numpy as np
```

```python
# Creating a Dataframe of random numbers with 100 rows and 5 columns.

dataframe = pd.DataFrame(np.random.rand(100, 5))
```

```python
# Printing the Dataframe

dataframe
```

```python
# this displays a specific part of the dataframe.

dataframe.loc[[10, 20], [1, 2]]
```

```python
# This shows all the columns of the dataframe but some specific rows.

dataframe.loc[[10, 20],:]
```

```python
# This shows all the rows of the dataframe but some specific columns.

dataframe.loc[:,[1,2]]
```

```python
# This gives all the rows along with their columns where the index is 0 for the row and the value of the data stored in that location of the table is less than 0.5. It is complex database query.

dataframe.loc[(dataframe[0] < 0.5)]
```

```python
# If the index of the table is something else like a alphabet then I want to access a particular element from the table then I have to give the specific index and I cannot use number if the index is not a number. This is the case for using .loc function, but we can use numbers even if the index is not a number using the .iloc function.

dataframe.iloc[0,4]
```

```python
# We have changed the indexing of the columns from numbers to alphabets.

dataframe.columns = list('ABCDE')
```

```python
# Printing the top five elements after changing the indexing of the columns.

dataframe.head()
```

```python
# Even after changing the index from numbers to alphabets we can still use numbers to access elements using the .iloc function.

dataframe.iloc[[0,1],[1,0]]
```

```python
# We are dropping the row of index 0, but this is temporarily done and original table is not changed in the memory.

dataframe.drop([0])
```

```python
# Printing the dataframe for testing the function of drop for row and we can see the drop function did not changed and worked as expected. It had changed the original tabel in the memory location if we have added a specific syntax to it.

dataframe
```

```python
# This is the additional syntax which the changes the table in the memory location permanently.

dataframe = dataframe.drop([0])
```

```python
# Printing the database to check for the expected functionality.

dataframe
```

```python
# This is another way to change the data in the table in the memory location permanently.

dataframe.drop([2], inplace=True)
```

```python
dataframe.head(4)
```

```python
# This is a complex query of an database which we are using here. This will print all the columns but not all the rows. This will check for the numbers in the columns 'C' and columns 'E' for numbers which are less than 0.9 and greater than 0.5 respectively and return those specific rows in which the specific index along with the column is present and print it.

dataframe.loc[(dataframe['C'] < 0.9) & (dataframe['E'] > 0.5)]
```

```python
# Printing the top five rows along with their columns to check for the indexin of the rows.

dataframe.head()
```

```python
# This resets the indexing of the rows to 0 to ... instead of 3 to ...(which is in this case), but the case is it does it temporarily and there is no change in the original dataframe located in the memory.

dataframe.reset_index()
```

```python
# Checking for the indexing of the row that it has changed or not.

dataframe
```

```python
# This will change the original dataframe as we have added the argument 'inplace' and the 'drop' argument removes the extra column along side the new indexing column. This extra column shows the original indexing of the dataframe before the change of the indexing. Also we can change the original dataframe using the old style like this 'dataframe = ...'. Also, it only changes the indexing of the rows and not the columns.

dataframe.reset_index(drop=True, inplace=True)
```

```python
# Printing the top five rows of the dataframe to check the intended behaviour of the 'reset_index' function of the dataframe.

dataframe.head()
```
