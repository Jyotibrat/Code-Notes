```python
# Creating a Dataframe consisting of 'names', 'toy' and 'born' and printing the dataframe.

df = pd.DataFrame({"name": ['Alfred', 'batman', 'Alfred'],
                        "toy": [np.nan, 'Batmobile', 'Bullwhip'],
                        "born": [pd.NaT, pd.Timestamp("1940-04-25"),
                        pd.NaT]})
df
```

```python
# This functions gives the count of the elements present in the column of the dataframe but by default it does not count the values like 'NaN' and 'NaT' but, if we put the argument 'dropna=False' then, it also gives the count of those elements too.

df['toy'].value_counts(dropna=False)
```

```python
# This function gives the count of elements and does count elements like 'NaN' and 'NaT'.

df['toy'].value_counts()
```

```python
# This shows all the value which are null in value in the dataset.

df.isnull()
```

```python
# It just gives the opposite result of the 'isnull' function.

df.notnull()
```
