```python
# Imported pandas and numpy library as pd and np respectively.

import pandas as pd
import numpy as np
```

```python
# Generating the integers from 0 to 100 in 10 rows and 10 columns and also printing the dataframe.

df = pd.DataFrame(np.random.randint(0, 100, size=(10, 10)))
df
```

```python
# This function makes dataframe to an excel sheet.

df.to_excel('Excel_Sheet.xlsx')
```

```python
# This function is used to read an excel sheet.

data = pd.read_excel('Excel_Sheet.xlsx')
data
```

```python
# Using this function we can write in a particular sheet in an excel sheet.

data_1 = pd.read_excel('Excel_Sheet.xlsx', sheet_name='Sheet1')
data_1
```

```python
# Generating another dataframe of integers from 100 to 200 in 10 rows and 10 columns.

df_2 = pd.DataFrame(np.random.randint(100, 200, size=(10, 10)))
df_2
```

```python
# We have copied the data and pasted in the excel sheet and then we are reading it that the data has changed in the sheet 2 of the excel sheet or not.

data_2 = pd.read_excel('Excel_Sheet.xlsx', sheet_name='Sheet2')
data_2
```

```python
# Changing the data at index [0][0], but here we are changing the copy and not not implemented in the original excel sheet which located somewhere in the memory.

data_2.iloc[0, 0] = 1000
data_2
```

```python
# Using this function we are writing in the excel sheet the data we have changed in the copy of the excel sheet.

data_2.to_excel('Excel_Sheet.xlsx', sheet_name='Sheet2')
```
